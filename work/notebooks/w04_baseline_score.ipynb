## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

Rule 1: 'Old content that hasn't been refreshed gets a lower ctr than other content'

Rule 2: “The better a page’s average search position, the higher its CTR.”

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
#Reason Code for Rule Number One.

r = df[df['gsc_impressions'] >= 100].copy()

def stale_group(d):
  if d['old'] == 0:
    return '1_fresh'
  return  '2_old_recent_update' if d['old_but_updated_recently'] == 1 else '3_old_not_updated'

r['stale_group'] = r.apply(stale_group, axis=1)

bucket_table_A = r.groupby('stale_group').agg(
    n=('ctr_fixed', 'size'),
    mean_ctr = ('ctr_fixed', 'mean'),
    median_ctr = ('ctr_fixed', 'median'),
    pct_zero_clicks = ('gsc_clicks', lambda s: ( s == 0 ).mean()),
)

bucket_table_A
```

Verdict:Mixed
1.Fresh pages have the highest CTR, but old-not-updated pages are only slightly behind (mean 0.0044 vs 0.0048; zero-click share 44% vs 43%), and old pages that were recently updated do worst (0.0038, 48% zero-click). Staleness alone separates CTR weakly and inconsistently, so verdict: MIXED.

```python
#Reason code for rule 2
r['pos_bucket'] = pd.cut(r['gsc_avg_position'], [0, 3, 10, 20 , 50, 100, np.inf])
bucket_table_b = r.groupby('pos_bucket', observed=True).agg(
    n=('ctr_fixed', 'size'),
    mean_ctr = ('ctr_fixed', 'mean'),
    median_ctr = ('ctr_fixed', 'median'),
    pct_zero_clicks = ('gsc_clicks', lambda s: ( s == 0 ).mean()),
)

bucket_table_b
```

Verdict:Confirmed
CTR falls monotonically as average position worsens: mean CTR drops from 0.0073 (positions 1-3) to 0.0007 (positions 51-100), and the zero-click share rises from 23% to 84%. Position clearly drives CTR, so comparing a page’s CTR to the norm for its position bucket is justified. Verdict: CONFIRMED. Caveat: the 51-100 bucket has only 339 rows, so its average is less reliable.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
d = df[df['gsc_impressions'] >= 100].copy()
d = d.sort_values('report_date')
d['pos_x_imp'] = d['gsc_avg_position'] * d['gsc_impressions']

pages = d.groupby('content_hash_id').agg(
    client=('client_hash_id_x', 'first'),
    content_type=('content_type', 'first'),
    old=('old', 'last'),
    old_but_updated_recently=('old_but_updated_recently', 'last'),
    impressions=('gsc_impressions', 'sum'),
    clicks=('gsc_clicks', 'sum'),
    pos_x_imp=('pos_x_imp', 'sum'),
    n_reports=('report_date', 'nunique'),
).reset_index()

pages['avg_position'] = pages['pos_x_imp'] / pages['impressions']
pages['ctr'] = pages['clicks'] / pages['impressions']
pages = pages[pages['impressions'] >= 100].copy()

pages['pos_bucket'] = pd.cut(pages['avg_position'], [0, 3, 5, 7, 10, 20, 50, 100])
g = pages.groupby('pos_bucket', observed=True)
pages['expected_ctr'] = g['clicks'].transform('sum') / g['impressions'].transform('sum')
pages['ctr_gap'] = (pages['expected_ctr'] - pages['ctr']).clip(lower=0)

flag = (pages['old'] == 1) & (pages['old_but_updated_recently'] == 0) & (pages['ctr_gap'] > 0)
pages['score'] = np.where(flag, pages['impressions'] * pages['ctr_gap'], 0)
pages['reason_code'] = np.where(flag, 'STALE_BELOW_POSITION_CTR', '')
pages['action'] = np.where(flag, 'REFRESH', 'NONE')

queue = pages[flag].sort_values('score', ascending=False)
```

```python
queue.to_csv('baseline_action_score.csv', index=False)
```
